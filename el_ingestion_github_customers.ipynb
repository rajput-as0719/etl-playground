**Library -> Module -> Functions**

```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv")
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
spark_df=spark.createDataFrame(df)
spark_df.write.format("delta").mode("overwrite").saveAsTable("batch_1.data1.customers")
```


### Task 

Learning ELT process: Spark Session 
- Import -> importing gives access to Spark’s APIs for creating sessions, manipulating DataFrames, and applying transformations.

- Session Creation -> It’s the entry point to programming with Spark using the DataFrame and SQL API.
- Dataframe Creation 
  - Dataframe? -> DataFrame is a distributed collection of data organized into named columns (like a table in SQL).
  - Data Write? -> write DataFrames back to storage in different formats

- Delta Format? 
  Delta Lake is an open-source storage layer that brings ACID transactions to Spark.
  
  It allows versioning, schema enforcement, and time travel (querying older snapshots of data).
  
  - Mode ? - Dalta format ensures transactional guarantees.
    -> overwrite → replaces existing data
    -> append → adds new data
    -> ignore → skips writing if data exists
    -> errorifexists → throws error if data exists

(ELT Flow)
-> Extract → Read raw data into Spark DataFrames (spark.read)

-> Load → Store data in Delta format for reliability (df.write.format("delta"))

-> Transform → Apply Spark SQL/DataFrame operations (df.select, df.filter, etc.)
